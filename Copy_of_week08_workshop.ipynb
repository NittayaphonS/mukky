## 🤖 AI Collaboration Exercise 1: Basic Analysis

**Your Turn**: Practice prompting AI for basic data analysis.

::: {.callout-tip}
## Practice Prompt Template

Try asking AI:
*"Using this transaction DataFrame, I want to understand my spending patterns. Please create code that shows:

**Expected AI Response** (try this yourself first!):

````python
# AI-generated solution for spending analysis

# Calculate comprehensive spending summary
spending_analysis = df.groupby('Category')['Amount'].agg({
    'Total_Spent': lambda x: pd.to_numeric(x.str.replace('$', ''), errors='coerce').sum(),
    'Avg_Amount': lambda x: pd.to_numeric(x.str.replace('$', ''), errors='coerce').mean(),
    'Transaction_Count': 'count'
}).round(2)

# Sort by total spending
spending_analysis = spending_analysis.sort_values('Total_Spent', ascending=False)

print("💼 Business Spending Analysis")
print("=" * 50)
print(f"{'Category':<15} {'Total':<10} {'Average':<10} {'Count':<8}")
print("-" * 50)

for category, row in spending_analysis.iterrows():
    total = row['Total_Spent']
    avg = row['Avg_Amount']
    count = row['Transaction_Count']
    print(f"{category:<15} ${total:<9.2f} ${avg:<9.2f} {count:<8.0f}")

print(f"\nGrand Total: ${spending_analysis['Total_Spent'].sum():.2f}")
````

````python
# AI-assisted data loading section
# (Copy the code AI provides here)
import pandas as pd

# 1. Load the CSV data
# We specify the file name 'transactions_2.csv' as requested.
file_path = 'transactions_2.csv'

try:
    df = pd.read_csv(file_path)
    print(f"✅ Successfully loaded '{file_path}'")
except FileNotFoundError:
    print(f"❌ Error: The file '{file_path}' was not found. Please ensure it is in the correct directory.")

# 2. Clean and Validate the Data
# The 'Amount' column often contains '$' and ',' which prevent mathematical calculations.
# We remove these characters and convert the column to a numeric float type.
df['Amount'] = df['Amount'].str.replace('$', '', regex=False).str.replace(',', '', regex=False)

# errors='coerce' is a critical safety feature: it turns unparseable strings into NaN (Not a Number)
# instead of crashing the program.
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

# Convert the 'Date' column to actual datetime objects to allow for chronological sorting and filtering.
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# 3. Handle Common Data Issues
# We identify missing values (NaN) which could result from empty cells or failed cleaning.
missing_summary = df.isnull().sum()

# We check for duplicate transactions, which is common in manual financial logging.
duplicates_count = df.duplicated().sum()

# 4. Show basic information about the dataset
print("\n" + "="*40)
print("💼 FINANCE ASSISTANT: INITIAL DATA REPORT")
print("="*40)

print("\n--- 1. Data Structure ---")
# .info() provides a high-level look at data types and memory usage.
df.info()

print("\n--- 2. Cleaned Data Preview (First 5 Rows) ---")
print(df.head())

print("\n--- 3. Data Integrity Summary ---")
print(f"Missing Values detected:\n{missing_summary}")
print(f"Duplicate Rows found: {duplicates_count}")

print("\n--- 4. Numerical Summary of Spending ---")
# .describe() provides mean, median, min, max, and standard deviation.
print(df['Amount'].describe())

# Add your own comments about what you learned
#there are some part that not neccessary to mention it such as this part (# We specify the file name 'transactions_2.csv' as requested.
#file_path = 'transactions_2.csv')
#should have a clear topic because when it provided longer details, it seems unprofessional.
````

## 📈 Step 2: Create Core Analysis Functions

## Analysis Functions Prompt:

*"For my Smart Finance Assistant, I need functions to analyze spending data. Please create well-documented functions that:
1. Calculate total spending by category
2. Find the top 5 spending categories
3. Calculate average spending per transaction
4. Identify any unusual transactions (very high or negative amounts)
Each function should have clear business explanations and error handling."*
:::

````python
# AI-generated analysis functions
# (Your AI collaboration code goes here)
import pandas as pd

def get_spending_by_category(df):
    """
    1. Calculate total spending per category.
    Business Explanation: This helps the organization visualize exactly where capital is being allocated
    to ensure spending aligns with strategic priorities.
    """
    try:
        # Group by Category and sum the cleaned Amount
        category_totals = df.groupby('Category')['Amount'].sum().sort_values(ascending=False)
        return category_totals
    except KeyError:
        return "Error: Required columns ('Category' or 'Amount') are missing from the dataset."
    except Exception as e:
        return f"An unexpected error occurred: {e}"

def get_top_5_categories(df):
    """
    2. Identify the top 5 spending categories.
    Business Explanation: Focusing on the top 5 cost drivers follows the Pareto Principle (80/20 rule),
    where managing a small number of categories can lead to significant overall savings.
    """
    try:
        # Sum spending and retrieve only the top 5 items
        top_5 = df.groupby('Category')['Amount'].sum().nlargest(5)
        return top_5
    except Exception as e:
        return f"Error identifying top 5 categories: {e}"

def get_average_transaction_value(df):
    """
    3. Calculate average spending per transaction.
    Business Explanation: Establishes a baseline for typical transaction sizes, allowing the business
    to monitor for inflation or changes in procurement behavior.
    """
    try:
        # Calculate the arithmetic mean of the Amount column
        avg_value = df['Amount'].mean()
        return round(avg_value, 2)
    except Exception as e:
        return f"Error calculating average spending: {e}"

def identify_unusual_transactions(df, high_threshold=1000):
    """
    4. Detect unusual transactions (very high or negative amounts).
    Business Explanation: Acts as an automated internal audit tool to flag potential fraudulent
    activity, accidental double-billing, or significant vendor refunds.
    """
    try:
        # Flag transactions exceeding a specific dollar amount
        high_spending = df[df['Amount'] > high_threshold]

        # Flag negative amounts which often represent refunds or bookkeeping errors
        negative_amounts = df[df['Amount'] < 0]

        return {
            "high_spending": high_spending,
            "negative_amounts": negative_amounts
        }
    except Exception as e:
        return f"Error identifying unusual transactions: {e}"

# --- Usage Example ---
# Replace 'df' with your actual cleaned DataFrame from Step 1

# print("--- Total Spending by Category ---")
# print(get_spending_by_category(df))

# print("\n--- Top 5 Spending Categories ---")
# print(get_top_5_categories(df))

# print(f"\n--- Average Transaction Value: ${get_average_transaction_value(df)} ---")

# # Example: Flagging transactions over $500 as 'High'
# alerts = identify_unusual_transactions(df, high_threshold=500)
# print(f"\n--- Alerts: Found {len(alerts['high_spending'])} high-value transactions ---")
````

## 💰 Step 3: Generate Business Insights

**Challenge AI** to help you create business-relevant insights:

::: {.callout-tip}
## Business Insights Prompt:

*"Using the transaction analysis functions, help me create a comprehensive spending report that would be useful for personal financial planning. Include recommendations for potential savings and spending pattern observations. Format it like a professional financial summary."*
:::

````python
# AI-generated business insights
# (Your AI collaboration code goes here)
# --- AI-Generated Business Insights & Strategic Report ---

def generate_business_insights(df, summary_report):
    """
    Analyzes summary data to generate automated strategic financial recommendations.
    """
    total_spent = df[df['Amount'] > 0]['Amount'].sum()
    insights = []

    # 1. Concentration Analysis (Pareto Principle)
    top_category = summary_report.index[0]
    top_percent = (summary_report.loc[top_category, 'sum'] / total_spent) * 100

    if top_percent > 50:
        insights.append(f"⚠️ **High Concentration Risk:** The '{top_category}' category accounts for {top_percent:.1f}% of total spending. "
                        f"Consider renegotiating contracts or auditing these large fixed costs.")

    # 2. Operational Pattern Analysis (Frequency vs. Value)
    most_frequent = df['Category'].value_counts().idxmax()
    freq_count = df['Category'].value_counts().max()

    insights.append(f"📊 **Operational Pattern:** High transaction frequency detected in '{most_frequent}' ({freq_count} times). "
                    f"Consolidating these purchases could reduce transaction fees and increase bargaining power.")

    # 3. Cash Recovery Analysis (Refunds/Credits)
    refunds = df[df['Amount'] < 0]['Amount'].abs().sum()
    if refunds > 0:
        insights.append(f"💰 **Cash Recovery:** A total of ${refunds:,.2f} in refunds/credits was detected. "
                        f"Ensure all pending returns are fully processed and credited to your account.")

    return insights

# --- Execution and Display ---

# Assuming 'summary' is the result from your Step 2 analysis
print("=" * 50)
print("💼 STRATEGIC BUSINESS INSIGHTS")
print("=" * 50)

# Generate insights
business_insights = generate_business_insights(df, summary)

# Print insights with professional formatting
for i, insight in enumerate(business_insights, 1):
    print(f"{i}. {insight}\n")

print("-" * 50)
print(f"Grand Total Expenditure: ${df[df['Amount'] > 0]['Amount'].sum():,.2f}")
print("-" * 50)
````
