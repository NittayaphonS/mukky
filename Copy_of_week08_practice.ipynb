## Business Context

## Practice Exercise 1.1: Sales Performance by Region

::: {.callout-tip}
## Your AI Collaboration Challenge

**Business Question**: "Which region is performing best, and what's the average sale size by region?"

**Your Task**: Practice prompting AI with this business context. Try creating a prompt that asks for:
1. Total sales by region
2. Average sale amount by region  
3. Number of transactions by region
4. Results formatted for a manager presentation

**Write your prompt below, then see what AI generates:**
:::

**Your Prompt:**
```
[I am working on a Sales Performance Analysis for a manager presentation. I have a dataset with 'Region', 'Amount', and 'Transaction_ID' columns.

Please write Python code using pandas to:

Calculate Total Sales by region.

Calculate the Average Sale Amount per region.

Count the Number of Transactions per region.]
```

**AI Response:**

````python
# Paste AI-generated code here and run it
# Remember to critique and improve if needed
import pandas as pd

# 1. Setup Sample Data (In a real scenario, you would use pd.read_csv('sales_data.csv'))
data = {
    'Region': ['North America', 'Europe', 'Asia-Pacific', 'North America', 'Europe',
               'Asia-Pacific', 'South America', 'North America', 'Europe', 'South America'],
    'Amount': [1200.50, 950.00, 1500.25, 800.00, 1100.75, 450.00, 300.00, 2100.00, 1050.00, 250.50],
    'Transaction_ID': range(101, 111)
}
df_sales = pd.DataFrame(data)

# 2. Core Analysis: Aggregating Sales Data by Region
# We calculate Sum, Mean, and Count in one go using .agg()
sales_performance = df_sales.groupby('Region')['Amount'].agg(
    Total_Sales='sum',
    Average_Sale='mean',
    Transaction_Count='count'
).sort_values(by='Total_Sales', ascending=False)

# 3. Formatting for Manager Presentation
# Making the data look professional with currency symbols and rounding
formatted_report = sales_performance.copy()
formatted_report['Total_Sales'] = formatted_report['Total_Sales'].apply(lambda x: f"${x:,.2f}")
formatted_report['Average_Sale'] = formatted_report['Average_Sale'].apply(lambda x: f"${x:,.2f}")

# 4. Display the Final Results
print("="*60)
print("📈 REGIONAL SALES PERFORMANCE REPORT")
print("="*60)
print(formatted_report)

# Identifying the Top Performer
top_region = sales_performance.index[0]
print(f"\n💡 Best Performing Region: {top_region}")
````

**Your Critique:**
- What did AI do well?

= Clear Insights: By sorting the results in descending order by Total_Sales, the AI made the answer to the business question obvious at a glance.

- What would you ask AI to improve?

= Adding a "Percentage of Total Sales" column would provide more strategic context regarding how much each region contributes to the whole company.

- Does the output help answer the business question?

= Yes. The output directly identifies the best-performing region at the top of the table and provides the average sale size requested by the business question.

## Practice Exercise 1.2: Top Performing Products

::: {.callout-tip}
## Business Scenario

Your inventory manager needs to know which products to stock more of next quarter.

**Challenge**: Ask AI to identify the top 5 products by:
- Total revenue generated
- Total units sold
- Which products have the highest profit margins (hint: compare unit price trends)

**Practice better prompting**: Be specific about formatting and business context.
:::

**Your Prompt:**
```
[I am acting as a Data Analyst for an Inventory Manager planning for the next quarter. Please write Python code using pandas to analyze a product dataset with the following columns: 'Product_Name', 'Units_Sold', 'Unit_Price', and 'Unit_Cost'.

I need you to generate three distinct reports to help with stocking decisions:
1. Top 5 Products by Total Revenue: (Units_Sold * Unit_Price). This identifies our high-value cash drivers.
2. Top 5 Products by Total Units Sold: This identifies our high-velocity items that require frequent restocking.
3. Top 5 Products by Profit Margin: Calculate the margin percentage using ((Unit_Price - Unit_Cost) / Unit_Price) * 100. This identifies our most efficient earners.

Output Requirements:
- Present each report as a separate, professionally formatted table.
- Use '$' for currency and '%' for margins, rounded to 2 decimal places.
- Include a brief business insight for each table explaining why these specific products should be prioritized for next quarter's inventory.]
```

**AI Response:**

````python
# AI-generated code goes here
import pandas as pd

# 1. Setup Sample Data
data = {
    'Product_Name': ['Ultra-Light Laptop', 'Mechanical Keyboard', '4K Monitor', 'USB-C Dock',
                    'Wireless Mouse', 'Noise-Cancelling Headphones', 'Desk Lamp', 'Ergonomic Chair'],
    'Units_Sold': [60, 240, 150, 500, 850, 200, 120, 45],
    'Unit_Price': [1200.00, 85.00, 350.00, 65.00, 25.00, 180.00, 35.00, 290.00],
    'Unit_Cost': [850.00, 40.00, 220.00, 20.00, 8.00, 110.00, 12.00, 210.00]
}
df = pd.DataFrame(data)

# 2. Calculations
df['Total_Revenue'] = df['Units_Sold'] * df['Unit_Price']
df['Profit_Margin'] = ((df['Unit_Price'] - df['Unit_Cost']) / df['Unit_Price']) * 100

# 3. Generating Reports
# A. Top 5 by Revenue
top_revenue = df.nlargest(5, 'Total_Revenue')[['Product_Name', 'Total_Revenue']]
top_revenue['Total_Revenue'] = top_revenue['Total_Revenue'].map("${:,.2f}".format)

# B. Top 5 by Units Sold
top_units = df.nlargest(5, 'Units_Sold')[['Product_Name', 'Units_Sold']]

# C. Top 5 by Profit Margin
top_margin = df.nlargest(5, 'Profit_Margin')[['Product_Name', 'Profit_Margin']]
top_margin['Profit_Margin'] = top_margin['Profit_Margin'].map("{:.2f}%".format)

# --- Final Presentation ---
print("📦 QUARTERLY INVENTORY STRATEGY REPORT")
print("="*55)

print("\n💰 [TOP 5 BY REVENUE] - Critical Cash Flow Drivers")
print("Insight: Stock these to ensure top-line growth targets are met.")
print(top_revenue.to_string(index=False))

print("\n📈 [TOP 5 BY UNITS SOLD] - High-Velocity Inventory")
print("Insight: These items move fast. Increase safety stock to avoid 'Out of Stock' events.")
print(top_units.to_string(index=False))

print("\n💎 [TOP 5 BY PROFIT MARGIN] - High-Efficiency Earners")
print("Insight: These yield the most profit per sale. Prioritize marketing for these items.")
print(top_margin.to_string(index=False))
````

**Business Insight Questions:**
- Which products should the company stock more of?

= Groceries , Coffee
- Are high-price items selling well, or is volume more important?

=volume is currently more important than high price. The Utilities category has the highest average price ($145.00), but it only occurred once however, Volume Analysis: Coffee and Transport have low average prices ($14.40 and $26.33 respectively) but account for nearly half of all transactions. These high-volume items provide the steady cash flow necessary to sustain operations.

- What would you recommend to the inventory manager?

= I recommend a Tiered Inventory Strategy:
- The Revenue Engine-Maintain high stock levels. Since this drives the most revenue, a "stock-out" here would be the most expensive mistake.
- The Traffic Driver-Ensure 100% availability. These items bring customers in frequently; use them as "loss leaders" to encourage purchases in higher-margin categories.
- The Selective Stock-Stock these carefully. They have a high average price ($86.65) and decent volume, making them great for boosting profit margins without the high storage overhead of groceries.

## Practice Exercise 2.2: Budget Category Analysis

::: {.callout-tip}
## Personal Finance Scenario

Someone wants to set a realistic monthly budget based on their spending history.

**Challenge**: Use AI to help:
1. Calculate average monthly spending by category
2. Identify categories with high variation (inconsistent spending)
3. Suggest realistic budget amounts for each category
4. Find categories where spending might be reduced

**Advanced**: Ask AI to explain the statistical concepts it uses.
:::

**Your Budget Analysis:** "Can you help me write Python code to analyze my spending data and create a monthly budget? Calculate average spending by category, find categories with inconsistent spending, suggest budget amounts, identify where I could cut back, and explain the statistics used."

````python
# AI-assisted budget recommendation code
import pandas as pd

# Load dataset
df = pd.read_csv("transactions.csv")

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Create Month column for monthly grouping
df['Month'] = df['Date'].dt.to_period('M')

# ============================================================================
# 1. Calculate Average Monthly Spending by Category
# ============================================================================
monthly_spending = df.groupby(['Month', 'Category'])['Amount'].sum().reset_index()

avg_monthly_spending = monthly_spending.groupby('Category')['Amount'].mean()

print("=== Average Monthly Spending by Category ===")
print(avg_monthly_spending)


# ============================================================================
# 2. Identify Categories with High Variation
# ============================================================================
variation = monthly_spending.groupby('Category')['Amount'].std()

print("\n=== Spending Variation by Category ===")
print(variation.sort_values(ascending=False))


# ============================================================================
# 3. Suggest Realistic Budget Amounts
# (Average + 10% Buffer)
# ============================================================================
suggested_budget = avg_monthly_spending * 1.10

print("\n=== Suggested Budget by Category ===")
print(suggested_budget)


# ============================================================================
# 4. Find Categories Where Spending Might Be Reduced
# (Top 3 Highest Spending Categories)
# ============================================================================
highest_spending = avg_monthly_spending.sort_values(ascending=False).head(3)

print("\n=== Categories to Review for Possible Reduction ===")
print(highest_spending)
````

## Scenario 2: Personal Finance Coaching

::: {.callout-note}
## Business Context

You're building a personal finance app feature that provides spending advice.

**Challenge**: Use AI to create code that:
1. Identifies spending categories that are above typical percentages
2. Suggests specific areas for potential savings
3. Provides encouraging feedback for good spending habits
4. Formats advice in a user-friendly way
:::

**Your Coaching Algorithm:**

````python
# AI-assisted personal finance advice generation

import pandas as pd

df = pd.read_csv("transactions.csv")

category_totals = df.groupby("Category")["Amount"].sum()
total_spending = category_totals.sum()

for category, amount in category_totals.items():
    percentage = (amount / total_spending) * 100

    if percentage > 30:
        print(f"{category}: High spending detected. Consider reducing expenses here.")
    else:
        print(f"{category}: Good spending habits in this category.")
````
